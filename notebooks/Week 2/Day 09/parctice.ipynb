```python
def count_up_to(n):
    i=1
    while i <= n:
        yield i
        i+=1

gen_func=count_up_to(5)
for num in gen_func:
    print(num)
```

This function is a Generator function 

```python
def shout(func):
    def greeting(m):
        print('before:', m)
        func() 
        print('after:', m.upper())  
    return greeting  

  
@shout
def middle_code():
    print("--- changing to uppercase ---")

middle_code("arham")
```

```python
def evens(limit):
    q=1
    while q<=limit:
        if q%2==0:
            yield q
        q+=1


for b in evens(10):
    print(b)
```
