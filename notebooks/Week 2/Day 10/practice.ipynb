```python
class student:
    def __init__(self,name,marks):
        self.name = name
        self.marks = marks

    def average(self):
        return sum (self.marks)/len(self.marks)
    
    def grade(self):

        avg = self.average()

        if avg < 70: return "C"
        
        elif avg < 80: return "B"
        elif avg < 100: return "A"
        elif avg > 100:return "Invalid marks"
        else:return "F"



ali=student("ali",[80,80,90,90])
print(ali.average())
print(ali.grade())
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"Refused: {self.owner} tried to withdraw Rs {amount} but has Rs {self.balance}")
            return
        self.balance -= amount
        

    def transfer(self,added,amount):

        self.owner=added
        if self.balance > amount:
            self.withdraw(amount)
            added.deposit(amount)
        else:
            print("Not efficient funds")

ali = BankAccount("Ali", 1000)
zara = BankAccount("Zara",500)
arham= BankAccount("Arham")

ali.transfer(zara,50)
print (zara.balance)
print (ali.balance)

ali.transfer(arham,450)

print(ali.balance)
print(arham.balance)


```
