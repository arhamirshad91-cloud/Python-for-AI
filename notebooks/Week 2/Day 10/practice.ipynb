```python
class student:
    def __init__(self,name,marks=[]):
        self.name = name
        self.marks = marks

    def average(self):
        return sum (self.marks)/len(self.marks)
    
    def grade(self):

        avg = self.average()

        if avg < 70: return "C"
        
        elif avg < 80: return "B"
        elif avg < 100: return "A"
        elif avg > 100:return "Invalid marks"
        else:return "F"



ali=student("ali",[80,80,90,90])
print(ali.average())
print(ali.grade())
```
